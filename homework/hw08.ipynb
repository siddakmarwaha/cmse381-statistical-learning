### Q9.7.8

a)

```python
OJ_data = pd.read_csv('OJ.csv')
OJ_data.head()
```

```python
OJ_data['STORE'].value_counts()
```

```python
OJ_data.shape[0]
```

```python
# Split the data
OJ_data_train, OJ_data_test = train_test_split(OJ_data, train_size=800, random_state=12, shuffle=True)
```

b)

```python
# Initialize label encoder
le = LabelEncoder()

categorical_columns = ['StoreID', 'SpecialCH', 'SpecialMM', 'Store7', 'STORE']

# Apply label encoding to categorical columns
for col in categorical_columns:
    OJ_data_train[col] = le.fit_transform(OJ_data_train[col])
    OJ_data_test[col] = le.transform(OJ_data_test[col])    

# Separate features and target
X_train = OJ_data_train.drop(columns=['Purchase'])
y_train = OJ_data_train['Purchase']

X_test = OJ_data_test.drop(columns=['Purchase'])
y_test = OJ_data_test['Purchase']

# Fit a support vector classifier
svc = SVC(C=0.01, kernel='linear')
svc.fit(X_train, y_train)

# Number of support vectors
num_support_vectors = len(svc.support_)
print("Number of support vectors:", num_support_vectors)
```

c)

```python
# Predict on training data
y_train_pred = svc.predict(X_train)
train_error = 1 - accuracy_score(y_train, y_train_pred)
print(f"Training error rate: {train_error}")

# Predict on test data
y_test_pred = svc.predict(X_test)
test_error = 1 - accuracy_score(y_test, y_test_pred)
print(f"Test error rate: {test_error}")
```

d)

```python
# Set up the parameter grid for C
param_grid = {'C': [0.01, 0.1, 1, 10]}

# Initialize GridSearchCV
grid_search = GridSearchCV(SVC(kernel='linear'), param_grid, cv=5)
grid_search.fit(X_train, y_train)

# Best C value found
best_C = grid_search.best_params_['C']
print(f"Best C from GridSearchCV: {best_C}")
```

e)

```python
# Fit the model with the best C
svc_optimal = SVC(C=best_C, kernel='linear')
svc_optimal.fit(X_train, y_train)

# Predict on training data
y_train_pred_optimal = svc_optimal.predict(X_train)
train_error_optimal = 1 - accuracy_score(y_train, y_train_pred_optimal)
print(f"Training error rate with optimal C: {train_error_optimal}")

# Predict on test data
y_test_pred_optimal = svc_optimal.predict(X_test)
test_error_optimal = 1 - accuracy_score(y_test, y_test_pred_optimal)
print(f"Test error rate with optimal C: {test_error_optimal}")
```

f)

```python
# Fit the SVC model with a radial kernel
svc_rbf = SVC(C=best_C, kernel='rbf')
svc_rbf.fit(X_train, y_train)

# Training and test error rates
y_train_pred_rbf = svc_rbf.predict(X_train)
train_error_rbf = 1 - accuracy_score(y_train, y_train_pred_rbf)
print(f"Training error rate with radial kernel: {train_error_rbf}")

y_test_pred_rbf = svc_rbf.predict(X_test)
test_error_rbf = 1 - accuracy_score(y_test, y_test_pred_rbf)
print(f"Test error rate with radial kernel: {test_error_rbf}")


# Set up the parameter grid for C
param_grid = {'C': [0.01, 0.1, 1, 10]}

# Initialize GridSearchCV
grid_search = GridSearchCV(SVC(kernel='rbf'), param_grid, cv=5)
grid_search.fit(X_train, y_train)

# Best C value found
best_C = grid_search.best_params_['C']
print(f"Best C from GridSearchCV: {best_C}")

# Fit the model with the best C found
svc_optimal = SVC(C=best_C, kernel='rbf')
svc_optimal.fit(X_train, y_train)

# Predict on training data
y_train_pred_optimal_rbf = svc_optimal.predict(X_train)
train_error_optimal_rbf = 1 - accuracy_score(y_train, y_train_pred_optimal_rbf)
print(f"Training error rate with optimal C: {train_error_optimal_rbf}")

# Predict on test data
y_test_pred_optimal_rbf = svc_optimal.predict(X_test)
test_error_optimal_rbf= 1 - accuracy_score(y_test, y_test_pred_optimal_rbf)
print(f"Test error rate with optimal C: {test_error_optimal_rbf}")
```

g)

```python
# Fit the SVC model with a polynomial kernel (degree=2)
svc_poly = SVC(C=best_C, kernel='poly', degree=2)
svc_poly.fit(X_train, y_train)

# Training and test error rates
y_train_pred_poly = svc_poly.predict(X_train)
train_error_poly = 1 - accuracy_score(y_train, y_train_pred_poly)
print(f"Training error rate with polynomial kernel: {train_error_poly}")

y_test_pred_poly = svc_poly.predict(X_test)
test_error_poly = 1 - accuracy_score(y_test, y_test_pred_poly)
print(f"Test error rate with polynomial kernel: {test_error_poly}")



# Set up the parameter grid for C
param_grid = {'C': [0.01, 0.1, 1, 10]}

# Initialize GridSearchCV
grid_search = GridSearchCV(SVC(kernel='poly', degree=2), param_grid, cv=5)
grid_search.fit(X_train, y_train)

# Best C value found
best_C = grid_search.best_params_['C']
print(f"Best C from GridSearchCV: {best_C}")

# Fit the model with the best C found
svc_optimal = SVC(C=best_C, kernel='poly', degree=2)
svc_optimal.fit(X_train, y_train)

# Predict on training data
y_train_pred_optimal_poly = svc_optimal.predict(X_train)
train_error_optimal_poly = 1 - accuracy_score(y_train, y_train_pred_optimal_poly)
print(f"Training error rate with optimal C: {train_error_optimal_poly}")

# Predict on test data
y_test_pred_optimal_poly = svc_optimal.predict(X_test)
test_error_optimal_poly= 1 - accuracy_score(y_test, y_test_pred_optimal_poly)
print(f"Test error rate with optimal C: {test_error_optimal_poly}")
```
